```python
df['Employment_Status']=df['Employment_Status'].replace({'EMP':'employed','Employed':'employed'})
```

```python
df['Employment_Status']=df['Employment_Status'].replace({'employed':1,
                                 'Unemployed':0,
                                 'Self-employed':2,
                                 'retired':3})
```

```python
df['Credit_Card_Type'] =df['Credit_Card_Type'].replace({'Student':0,'Platinum':1,'Gold':2,'Business':3,'Standard':4})
```

```python
month_cols = ['Month_1', 'Month_2', 'Month_3', 'Month_4', 'Month_5', 'Month_6']
```

```python
mapping_dict = {
    'Missed': 0,
    'Late': 1,
    'On-time': 2}
```

```python
df[month_cols] = df[month_cols].replace(mapping_dict)
```

```python
df['Income']=df['Income'].fillna(0).round().astype(int)
```

```python
abc= df['Credit_Score'].mean()
df['Credit_Score']=df['Credit_Score'].fillna(abc).round()
df['Credit_Score']=df['Credit_Score'].astype(int)
```

```python
df['Credit_Score']= df['Credit_Score'].astype(int)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
```

```python
y = df.iloc[:, 6]
x = df.iloc[:, [i for i in range(19) if i not in (0, 12, 6)]]
```

```python
x
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
```

```python
modal = DecisionTreeClassifier
```

```python
modal = DecisionTreeClassifier()
modal.fit(x_train, y_train)
```

```python
y_predict = modal.predict(x_test)
```

```python
print(f"Accuracy Score: {accuracy_score(y_test, y_predict)}")
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
R_modal=RandomForestClassifier(n_estimators=100,random_state=1)
R_modal.fit(x_train,y_train)
```

```python
y_predicts = R_modal.predict(x_test)
```

```python
print(f"Accuracy Score: {accuracy_score(y_test, y_predicts)}")
```
