```python
import pandas as pd
df = pd.read_csv('/content/amazon[1].csv')
```

```python
df
```

```python
df_copy = df[['product_name','category', 'discounted_price','actual_price','discount_percentage', 'rating','rating_count']]
```

```python
df_copy
```

```python
df_copy.info()
```

```python
import pandas as pd


df['main_category'] = df['category'].astype(str).str.split('|').str[0]

print(df['main_category'].value_counts().head(10))

```

```python
df_copy['category']
```

```python
df_copy.describe(include='all')
```

```python
df_copy['rating_count'].unique()
df_copy['rating_count'] = df_copy['rating_count'].str.replace(',', '').str.replace(' ', '').astype(float)
```

```python
median = pd.to_numeric(df['rating_count'].str.replace(',', '').str.replace(' ', ''), errors='coerce').median()
df_copy['rating_count'] = pd.to_numeric(df_copy['rating_count'], errors='coerce').fillna(median)
```

```python
df_copy['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)
```

```python
df_copy['discounted_price'] = pd.to_numeric(df_copy['discounted_price'].astype(str).str.replace('₹', '').str.replace(',', ''), errors='coerce')
```

```python
df_copy['rating'] = pd.to_numeric(df_copy['rating'].astype(str).str.replace('|', '2', regex=False), errors='coerce')
```

```python
df_copy['discounted_price'] = df_copy['discounted_price'].astype(str).str.replace('₹', '').str.replace(',', '').astype(float)
```

```python
top_discount = df_copy.groupby('category')['discounted_price'].mean().sort_values(ascending=False)
```

```python
top_discount.head(10)
```

```python
correlation = df_copy[['discounted_price', 'rating']].corr()
```

```python
correlation
```

```python
median=df_copy['actual_price'].median()
```

```python
expensive_product_av_rating = df_copy[df_copy['actual_price'] >median]['rating'].mean()
cheap_product_av_rating = df_copy[df_copy['actual_price'] <= median]['rating'].mean()
```

```python
print('expensive_product_av_rating', expensive_product_av_rating)
print('cheap_product_av_rating', cheap_product_av_rating)
```

```python
most_popular_products=df_copy.groupby('product_name')[['rating_count','rating']].mean().sort_values(by ='rating_count',ascending=False)
```

```python
most_popular_products.head(10)
```

```python
low_count_highRated_products = df_copy[df_copy['rating_count'] < 50][['product_name', 'rating_count', 'rating']].sort_values(by='rating', ascending=False)
low_count_highRated_products.head(10)
```

```python
market_dominance = df_copy.groupby('category')['product_name'].count().sort_values(ascending=False)
market_dominance.head(10)
```
